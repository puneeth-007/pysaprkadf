## Initialize PySpark Session and Load Data

Now, let's initialize a SparkSession and load the downloaded `train.csv` file into a PySpark DataFrame.

```python
# Initialize SparkSession
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("TitanicAnalysis").getOrCreate()

print("SparkSession initialized.")
```

```python
# Load the dataset into a PySpark DataFrame
titanic_df = spark.read.csv("titanic_data/train_and_test2.csv", header=True, inferSchema=True)

print("Dataset loaded successfully.")

# Display schema and first 5 rows to verify
titanic_df.printSchema()
titanic_df.show(5)
```
